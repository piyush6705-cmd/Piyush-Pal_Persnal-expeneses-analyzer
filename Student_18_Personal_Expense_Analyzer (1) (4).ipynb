## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_18_dataset.csv")
display(df)
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python

print("Shape:", df.shape)
print("Columns:", df.columns)

# Display data types
print("Data Types:")
print(df.dtypes)

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Count duplicate rows
print("Duplicate Rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
# Select numeric column
numeric_data = df[["Amount"]]

# Convert into NumPy array
amounts = numeric_data.to_numpy()

# Calculate average, maximum and minimum
average = np.mean(amounts)
maximum = np.max(amounts)
minimum = np.min(amounts)

print("Average Expense:", average)
print("Maximum Expense:", maximum)
print("Minimum Expense:", minimum)
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
# Create calculated column
df["Month"] = pd.to_datetime(df["Date"]).dt.month

df["Expense_Level"] = np.where(
    df["Amount"] >= 300, "High", "Normal"
)

# Filter records
filtered_data = df[df["Amount"] >= 300]

print("Filtered Expenses:")
display(filtered_data)

# Sort records
sorted_data = df.sort_values(
    by="Amount", ascending=False
)

print("Top Expenses:")
display(sorted_data.head(10))
```

## 6. Your individual project challenge
**Challenge:** Calculate total expenses; category-wise totals and averages; identify largest expense category.

Write your solution below and explain what the result means.


```python

total_expenses = df["Amount"].sum()

# Category-wise total and average
category_summary = df.groupby("Category")["Amount"].agg(
    ["sum", "mean"]
)

# Rename columns
category_summary.columns = [
    "Total Expenses", "Average Expenses"
]

# Display results
print("Total Expenses:", total_expenses)

print("Category Wise Analysis:")
display(category_summary)

# Identify largest expense category
largest_category = category_summary[
    "Total Expenses"
].idxmax()

print("Largest Expense Category:", largest_category)
```

## 7. Export your report
Use `to_csv(..., index=False)` to save your cleaned or analysed DataFrame. Create an `output` folder if needed.


```python
# TODO: Save your final report as a CSV file
category_summary.to_csv(
    "Student_18_expense_analysis.csv"
)

print("Report Saved Successfully!")
```
